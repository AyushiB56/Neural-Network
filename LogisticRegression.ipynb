Prevent Data Leakage:
If you scale before splitting, information from the test set leaks into the scaling process for the train set.

Best Practice:
Fit the scaler only on the training data, then apply the same transformation to the test data. This ensures your test data simulates “unseen” real-world data.

```python
X_train, x_test, y_train,y_test = train_test_split(df,target, test_size=0.2, random_state= 42 )
```
